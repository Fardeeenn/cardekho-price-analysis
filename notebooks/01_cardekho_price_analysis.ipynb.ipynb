## Univariate Analysis — Key Insights

- **Vehicle Age:** Most cars are around **5–7 years old**.
- **KM Driven:** Most vehicles have moderate usage, with some high-mileage cars.
- **Engine & Max Power:** Values are concentrated around common passenger-car ranges, with a few higher values.
- **Mileage:** Most cars fall within a typical mileage range.
- **Selling Price:** Prices are **right-skewed**, with a small number of expensive cars.
- **Seats:** Most cars have typical seating capacities; suspicious values were investigated.
- **Brand:** **Maruti, Hyundai, Honda, and Mahindra** are among the most represented brands.
- **Categorical Variables:** A few seller, fuel, and transmission categories dominate the dataset.
- **Outliers:** Potential outliers were investigated individually rather than automatically removed.

### Overall Takeaway
Univariate analysis helped understand the distribution of each variable and identify unusual values before moving to deeper analysis.

## **BRAND ANALYSIS**

- **BMW** had the highest average selling price among the 10 most represented brands.

- Average selling prices varied substantially across the major brands, suggesting **brand is an important factor** associated with used-car prices.

- **BMW's** higher average price was not simply due to having newer vehicles, as its average vehicle age was around **6.6 years**.

- **Tata** and **Renault** had relatively newer vehicles on average but substantially lower average selling prices than BMW.

> **NOTE:** For price and vehicle-age comparisons, the analysis was restricted to the **10 most represented brands** to avoid conclusions being driven by brands with very few observations.
